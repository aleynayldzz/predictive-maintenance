# Adim 1 - Veri Yukleme ve RUL Hesaplama

```python
import pandas as pd
import numpy as np
```

```python
column_names = ['unit_id', 'time_cycle', 'op_setting_1', 'op_setting_2', 'op_setting_3'] + [f'sensor_{i}' for i in range(1, 22)]

df = pd.read_csv(
    "data/train_FD001.txt",
    sep='\s+',
    header=None,
    names=column_names
)
```

```python
df.shape
```

```python
max_cycle = df.groupby('unit_id')['time_cycle'].transform('max')
```

```python
df['RUL'] = max_cycle - df['time_cycle']
```

```python
df[['unit_id', 'time_cycle', 'RUL']].head(10)
```

```python
df.isnull().sum()
```

```python
sensor_cols = [f'sensor_{i}' for i in range(1, 22)]
df[sensor_cols].std()
```

```python
threshold = 0.01
constant_sensors = df[sensor_cols].std()[df[sensor_cols].std() < threshold].index
constant_sensors
```
